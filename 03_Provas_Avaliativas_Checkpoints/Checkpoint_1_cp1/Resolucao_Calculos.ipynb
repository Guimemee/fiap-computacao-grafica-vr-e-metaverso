> Desenvolver shaders GLSL customizados para simular o modelo de iluminação de Phong (Componentes Ambiente, Difusa e Especular) sobre malhas poligonais 3D.

### Formulação Matemática Formal
$$I = I_a k_a + I_d k_d (\mathbf{L} \cdot \mathbf{N}) + I_s k_s (\mathbf{R} \cdot \mathbf{V})^\alpha$$


### Passo 1: Transformação de Coordenadas de Vértices
Aplicação das matrizes Model, View e Projection (MVP) no Vertex Shader.


### Passo 2: Cálculo da Componente Difusa (Lambertiana)
Produto escalar saturado \max(0, \mathbf{L} \cdot \mathbf{N}) entre vetor de luz e normal da superfície.


### Passo 3: Cálculo do Brilho Especular de Phong
Reflexão \mathbf{R} = 2(\mathbf{L} \cdot \mathbf{N})\mathbf{N} - \mathbf{L} e produto com vetor de visão \mathbf{V} elevado ao expoente de rugosidade \alpha.


### Passo 4: Interpolação no Fragment Shader
Rasterização e interpolação baricêntrica suave das cores em cada pixel da tela.


### Passo 5: Conclusão Visual
Renderização fotorrealista em tempo real a 120 FPS no WebGL.


```python
import numpy as np
import matplotlib.pyplot as plt

angulo = np.linspace(-90, 90, 100)
rad = np.radians(angulo)
difusa = np.maximum(0, np.cos(rad))
especular_10 = np.maximum(0, np.cos(rad))**10
especular_50 = np.maximum(0, np.cos(rad))**50
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#1c0d29')
ax.set_facecolor('#1c0d29')
ax.plot(angulo, difusa, color='#c084fc', lw=2, label='Difusa (Lambert)')
ax.plot(angulo, especular_10, color='#f472b6', lw=2, linestyle='--', label='Especular (α = 10)')
ax.plot(angulo, especular_50, color='#ffffff', lw=2.5, label='Especular (α = 50)')
ax.set_title('Modelo de Iluminação de Phong: Intensidade vs Ângulo', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Ângulo de Incidência da Luz (graus)', color='#94a3b8')
ax.set_ylabel('Intensidade Luminosa Relativa', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#432161', alpha=0.5)
ax.legend(facecolor='#2d1642', edgecolor='#c084fc', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
