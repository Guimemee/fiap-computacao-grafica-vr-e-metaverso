```python
import numpy as np
import matplotlib.pyplot as plt

cenarios = ['Lightmaps Tradicionais (Estático)', 'Unreal Engine 5 (Lumen & Nanite)']
tempo_build_min = [180, 0]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#1c0d29')
ax.set_facecolor('#1c0d29')
bars = ax.bar(cenarios, tempo_build_min, color=['#ef4444', '#10b981'], width=0.4)
ax.set_title('Tempo de Build de Iluminação: Tradicional vs UE5 Lumen', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Tempo de Compilação (minutos)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#432161', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 4, f'{yval} min', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
